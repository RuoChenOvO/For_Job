```python
import torch
print(torch.__version__)
print(torch.cuda.is_available()) # 返回True就是GPU可用
print(torch.version.cuda)
```

```python
x = torch.rand(4, 3) 
print(x)
```

```python
x = torch.zeros(4, 3, dtype=torch.long)
print(x)
```

```python
x = torch.tensor([5.5, 3]) 
print(x)
```

```python
x = torch.ones(4, 3, dtype=torch.double) 
# 创建一个新的全1矩阵tensor，返回的tensor默认具有相同的torch.dtype和torch.device
# 也可以像之前的写法 x = torch.ones(4, 3, dtype=torch.double)
print(x)
x = torch.randn_like(x, dtype=torch.float)
# 重置数据类型
print(x)
# 结果会有一样的size
# 获取它的维度信息
print(x.size())
print(x.shape)
```

```python
y = torch.rand(4, 3) 
print(y)
print(x + y)

# 方式2
print(torch.add(x, y))

# 方式3 in-place，原值修改
y.add_(x) 
print(y)
```

```python
z=y.unsqueeze(1)
print(z)
print(z.shape)
```
