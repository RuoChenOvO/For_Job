```python
# 空列表
lst1 = []
lst2 = list()

# 带元素
lst = [10, 20, 30, "hello", True]
```

```python
lst = [10,20,30]
print(lst[0])   # 10  第1个元素
print(lst[1])   # 20
print(lst[-1])  # 30  -1代表最后一个元素
```

```python
lst = [0,1,2,3,4,5]
print(lst[1:4])   # [1,2,3]  取下标1,2,3
print(lst[:3])    # [0,1,2]  从头到下标3（不含）
print(lst[2:])    # [2,3,4,5] 下标2到末尾
print(lst[::2])   # [0,2,4] 步长2，隔一个取一个
print(lst[::-1])  # [5,4,3,2,1,0] 反转列表
```

```python
lst = [99]
lst.append(100)
lst.append(1)
print(lst) # [99,100,1]
lst = [1]
lst.append([2,3])
print(lst) # [1, [2,3]]  注意：不是 [1,2,3]
```

```python
# extend () 批量追加多个元素，把另一个列表拆开加进去
lst = [1,2]
lst.extend([3,4])
print(lst) # [1,2,3,4]
```

```python
# insert() 在指定位置插入元素
lst = [1,3]
lst.insert(1,2)
print(lst) # [1,2,3]
```

```python
lst = [10,20]
lst[0] = 99
print(lst) # [99, 20]
```

```python
lst = [10,20,30,0]
lst.remove(20)     # 删除第一个值等于20的元素
print(lst) # [10, 30]
lst.pop()          # 默认删掉最后一个，返回被删掉的值
print(lst) # [10]
lst.pop(0)         # 删除下标0的元素
print(lst) # []
del lst[0]         # 删除下标0元素
print(lst) # []
lst.clear()        # 清空列表，变成[]
```

```python
# 常用查询操作
lst = [10,20,30]
print(20 in lst)        # True，判断元素是否存在
print(lst.index(20))    # 1，返回元素第一次出现的下标
print(lst.count(10))    # 1，统计这个元素出现多少次
print(len(lst))         # 3，获取列表长度
```

```python
lst = [10,20,30]
# 方式1：直接遍历元素
for item in lst:
    print(item)

# 方式2：带下标遍历
for idx,item in enumerate(lst):
    print(idx, item)
```

```python
a = [1,2]
b = [3,4]
print(a + b) # [1,2,3,4] 拼接
print(a * 2) # [1,2,1,2] 复制两遍
```
