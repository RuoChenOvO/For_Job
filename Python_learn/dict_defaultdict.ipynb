```python
d = dict()
# d = {}
d['a'] = 1
d['b'] = 2
print(d['a'])  # 1
print(d)  # {'a': 1, 'b': 2}
d['b'] += 1
print(d['b'])  # {'a': 1, 'b': 3}
```

```python
# 删除元素
d = {"a":1,"b":2,"c":3}

# pop：删除指定key，返回被删掉的值
val = d.pop("a")
print(val) # 1

# del 删除
del d["b"]
print(d) # {'c': 3}

# clear 清空整个字典
d.clear()
print(d) # {}
```

```python
d = {"a":1, "b":2}

# .keys() 拿到所有的键
print(d.keys())    # dict_keys(['a', 'b'])

# .values() 拿到所有的值（isAnagram里面用到了！）
print(d.values())  # dict_values([1, 2])

# .items() 同时拿到key和value
print(d.items())   # dict_items([('a', 1), ('b', 2)])
```

```python
d = {}
d['b'] = d.get('b', 0) + 1
# 第一次：get拿到0 → 0+1=1，存入d['b']
d['a'] = d.get('a', 0) + 1
print(d['b']) # 输出1
```

```python
for k in d:           # 遍历所有key
    print(k, d[k])

for v in d.values():  # 只遍历value
    print(v)

for k,v in d.items(): # key和value一起拿
    print(k,v)
```

```python
from collections import defaultdict

# defaultdict(类型)
# int → 默认值是0；list默认是空列表[]
d = defaultdict(int)

d['b'] += 1   # ✅ 不会报错！
# 流程：发现'b'不存在 → 自动创建 d['b'] = 0 → 0+1=1
print(d['b']) # 1
```

```python
from collections import defaultdict

d = defaultdict(list)
d['c'].append(99)
print(d['c']) # [99]
```

```python
for k in d:
    print(k, d[k])

for v in d.values():
    print(v)

for k,v in d.items():
    print(k,v)
```

```python
# 判断 key 是否在字典里面
d = {"a":1}
print("a" in d)   # True
print("c" in d)   # False
```

```python
# 合并字典
d1 = {"a":1}
d2 = {"b":2}
d3 = {**d1,**d2}
print(d3) # {'a':1, 'b':2}
```

```python
# 字典是可变对象，赋值是引用（和 list 一样）
d1 = {"a":1}
d2 = d1
d2["a"] = 99
print(d1) # {'a':99} d1跟着变
```

```python
lst = ['a','e','t']

res1 = "".join(lst)   # 分隔符是空，直接连起来 → "aet"
res2 = "-".join(lst)  # 分隔符是横线 → "a-e-t"

print(res1) # aet
print(res2) # a-e-t
```
